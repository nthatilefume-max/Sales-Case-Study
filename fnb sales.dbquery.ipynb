```sql
select * from `sales`.`case`.`study` limit 100;

-------Checking daily sales per unit
SELECT
    Date,
    Sales,
    `Quantity Sold`,
    ROUND(Sales / `Quantity Sold`, 2) AS Daily_Unit_Price
FROM `sales`.`case`.`study`
ORDER BY Date; 

--------Average unit sales price
SELECT
    ROUND(AVG(Sales / `Quantity Sold`), 2) AS Average_Unit_Price
FROM `sales`.`case`.`study`;

--------average monthly unit price
SELECT
    MONTH(Date) AS Month,
    ROUND(AVG(Sales / `Quantity Sold`), 2) AS Average_Unit_Price
FROM `sales`.`case`.`study` 
GROUP BY MONTH(Date)
ORDER BY Month;

--------Daily gross profit
SELECT
    Date,
    Sales,
    `Cost Of Sales`,
    (Sales - `Cost Of Sales`) AS Gross_Profit
FROM `sales`.`case`.`study`
ORDER BY Date;

-------daily gross profit percentage
SELECT
    Date,
    ROUND(((Sales - `Cost Of Sales`) / Sales) * 100,2) AS Gross_Profit_Percentage
FROM `sales`.`case`.`study`
ORDER BY Date;

-------highest sales day

SELECT
    Date,
    Sales,
    `Quantity Sold`
FROM `sales`.`case`.`study`
ORDER BY Sales DESC
LIMIT 1;

-------lowest sales day
SELECT Date,
       Sales
FROM `sales`.`case`.`study`
ORDER BY Sales ASC
LIMIT 1;

-----monthly sales performance
SELECT
    MONTH(Date) AS Month,
    SUM(Sales) AS Total_Sales,
    SUM(`Quantity Sold`) AS Total_Quantity,
    SUM(Sales - `Cost Of Sales`) AS Total_Profit
FROM `sales`.`case`.`study`
GROUP BY MONTH(Date)
ORDER BY Month;

------Top 10 best selling days
SELECT Date,
       Sales,
       `Quantity Sold`
FROM `sales`.`case`.`study`
ORDER BY Sales DESC
LIMIT 10;

------Running total sales
SELECT Date,
       Sales,
    SUM(Sales) OVER (ORDER BY Date) AS Running_Total_Sales
FROM `sales`.`case`.`study`;

------Running profit
SELECT Date,
       (Sales - `Cost Of Sales`) AS Profit,

       SUM(Sales - `Cost Of Sales`) OVER(ORDER BY Date) AS Running_Profit
FROM `sales`.`case`.`study`;

-------sales performance
SELECT
    Date,
    Sales,

    CASE
        WHEN Sales >= 10000 THEN 'Excellent'
        WHEN Sales >= 7000 THEN 'Good'
        WHEN Sales >= 4000 THEN 'Average'
        ELSE 'Low'
    END AS Sales_Performance
FROM `sales`.`case`.`study`
ORDER BY Sales DESC;

-------average daily sales
SELECT
    ROUND(AVG(Sales),2) AS Average_Daily_Sales,
    ROUND(AVG(`Quantity Sold`),2) AS Average_Daily_Quantity
FROM `sales`.`case`.`study`;

----promotion detection
SELECT
    Date,
    Sales,
    `Quantity Sold`,
    ROUND(Sales / `Quantity Sold`,2) AS Unit_Price,
    CASE
        WHEN (Sales / `Quantity Sold`) < (SELECT AVG(Sales / `Quantity Sold`) FROM `sales`.`case`.`study`)
        THEN 'Promotion'
        ELSE 'Normal Price'
    END AS Promotion_Status
FROM `sales`.`case`.`study`
ORDER BY Date;

-------daily KPI report
SELECT
    Date,
    Sales,
    `Cost Of Sales`,
    `Quantity Sold`,
    ROUND(Sales / `Quantity Sold`,2) AS Unit_Price,
    ROUND((Sales - `Cost Of Sales`),2) AS Gross_Profit,
    ROUND(((Sales - `Cost Of Sales`)/Sales)*100,2) AS Gross_Profit_Percentage,
    ROUND((Sales - `Cost Of Sales`)/`Quantity Sold`,2) AS Profit_Per_Unit
FROM `sales`.`case`.`study`
ORDER BY Date;
```
